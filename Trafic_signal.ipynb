```python
import time
print("Let's Start Traffic Signal : Press - '1'")
print("Let's Stop Traffic Signal  : Press - '2'")

def time_run():

  for _ in range(2):
    print("\n")
    print(">>>>>>--TRAFFIC--SIGNAL--<<<<<<")
    for i in range(1,16): #--> we change time in this loop
      #print(f"\r{i}", end="") --> the number rewrite in smae place
      if i <=5:
        print(f"\r  {i} - 🟡 <<READY>>",end="",flush=True)
      elif i <=10:
        print(f"\r  {i} - 🟢 <<GO>>",end="",flush=True)
      elif i <=15:
        print(f"\r  {i} - 🔴 <<STOP>>",end="",flush=True)
      else:
        print("Invalid Number")
      time.sleep(1)

while True:

  choose =int(input("\nEnter your choice :"))
  if choose == 1:
    time_run()
  elif choose == 2:
    print("Power off")
    break
  else:
    print("Invalid inputs")
```
